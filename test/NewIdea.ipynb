# Feature Engineering with Neural Network Models on Financial Indicators

This document describes the process of generating new feature representations of financial indicators using various neural network models. Each model will be applied to specific indicators, generating transformed versions of each indicator (e.g., `MACD_CNN`, `ADX_LSTM`) to capture different patterns and perspectives within the data.

## Objectives

1. **Leverage Neural Networks**: Use deep learning models to extract insights from financial indicators by generating transformed versions of each indicator.
2. **Enhanced Feature Set**: Create a comprehensive dataset with additional features for better prediction accuracy on tasks like price direction or trend forecasting.

## Approach

For each indicator, we’ll apply a series of neural network models, which include:

- Convolutional Neural Networks (CNNs)
- Long Short-Term Memory Networks (LSTMs)
- Recurrent Neural Networks (RNNs)
- Generative Adversarial Networks (GANs)
- Radial Basis Function Networks (RBFNs)
- Multilayer Perceptrons (MLPs)
- Self-Organizing Maps (SOMs)
- Deep Belief Networks (DBNs)

Each transformed feature is created by training the respective neural network on the indicator’s time-series data, capturing trends, dependencies, or patterns that are not evident from the raw indicator alone.

## Data Preparation

1. **Indicator Calculation**: Begin by calculating standard financial indicators (e.g., MACD, RSI, ADX) on your dataset.
2. **Normalization**: Normalize each indicator to ensure consistent scale across inputs.
3. **Windowing**: For sequence-based models (CNNs, LSTMs, RNNs), create sliding windows of each indicator to provide a sequential context.

## Model Definitions

Each model is defined to capture unique perspectives on the indicator’s behavior:

### 1. Convolutional Neural Network (CNN)

**Purpose**: Detect short-term trends or patterns within each indicator series.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, Flatten, Dense

def build_cnn(input_shape):
    model = Sequential([
        Conv1D(filters=32, kernel_size=3, activation='relu', input_shape=input_shape),
        Flatten(),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model
```

### 2. Long Short-Term Memory Network (LSTM)

**Purpose**: Capture long-term temporal dependencies within the indicator series.

```python
from tensorflow.keras.layers import LSTM

def build_lstm(input_shape):
    model = Sequential([
        LSTM(50, activation='relu', input_shape=input_shape),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model
```

### 3. Recurrent Neural Network (RNN)

**Purpose**: Learn sequential patterns that can predict future indicator movements.

```python
from tensorflow.keras.layers import SimpleRNN

def build_rnn(input_shape):
    model = Sequential([
        SimpleRNN(50, activation='relu', input_shape=input_shape),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model
```

### 4. Multilayer Perceptron (MLP)

**Purpose**: Fit general trends in the indicator data.

```python
from tensorflow.keras.layers import Dense

def build_mlp(input_shape):
    model = Sequential([
        Dense(64, activation='relu', input_shape=input_shape),
        Dense(32, activation='relu'),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model
```

### 5. Self-Organizing Map (SOM)

**Purpose**: Cluster similar patterns in the indicator series.

```python
from minisom import MiniSom

def build_som(data, x_dim=10, y_dim=10):
    som = MiniSom(x=x_dim, y=y_dim, input_len=data.shape[1], sigma=1.0, learning_rate=0.5)
    som.random_weights_init(data)
    return som
```

### 7. Radial Basis Function Network (RBFN) – Placeholder

### 8. Deep Belief Network (DBN) – Placeholder

## Main Feature Generation Pipeline

### Prepare Data for Windowed Models

1. **Normalization**: Scale each indicator to a range of 0-1.
2. **Sliding Windows**: Generate windows of data to provide context for each model.

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Normalize indicators
scaler = MinMaxScaler()
df_scaled = scaler.fit_transform(df)

# Create sliding windows
window_size = 10
X_windows = []
for i in range(len(df_scaled) - window_size + 1):
    window = df_scaled[i:i + window_size]
    X_windows.append(window)
X_windows = np.array(X_windows)
```

### Apply Models to Each Indicator

The following function loops over each indicator, applies each model, and stores the generated features:

```python
def create_indicator_features(df, indicators, window_size=10):
    X_windows = []
    new_features = pd.DataFrame(index=df.index)

    for indicator in indicators:
        data = df[[indicator]].values
        windows = [data[i:i + window_size] for i in range(len(data) - window_size + 1)]
        windows = np.array(windows)

        # CNN
        cnn_model = build_cnn((window_size, 1))
        cnn_model.fit(windows, df[indicator][window_size-1:], epochs=5, verbose=0)
        cnn_features = cnn_model.predict(windows)
        new_features[f"{indicator}_CNN"] = np.pad(cnn_features.flatten(), (window_size-1, 0), 'constant')

        # LSTM
        lstm_model = build_lstm((window_size, 1))
        lstm_model.fit(windows, df[indicator][window_size-1:], epochs=5, verbose=0)
        lstm_features = lstm_model.predict(windows)
        new_features[f"{indicator}_LSTM"] = np.pad(lstm_features.flatten(), (window_size-1, 0), 'constant')

        # RNN
        rnn_model = build_rnn((window_size, 1))
        rnn_model.fit(windows, df[indicator][window_size-1:], epochs=5, verbose=0)
        rnn_features = rnn_model.predict(windows)
        new_features[f"{indicator}_RNN"] = np.pad(rnn_features.flatten(), (window_size-1, 0), 'constant')

        # MLP
        mlp_model = build_mlp((window_size,))
        windows_flat = windows.reshape(windows.shape[0], -1)
        mlp_model.fit(windows_flat, df[indicator][window_size-1:], epochs=5, verbose=0)
        mlp_features = mlp_model.predict(windows_flat)
        new_features[f"{indicator}_MLP"] = np.pad(mlp_features.flatten(), (window_size-1, 0), 'constant')

        # SOM
        som_model = build_som(windows_flat)
        som_model.train_random(windows_flat, 100)
        som_features = np.array([som_model.winner(x) for x in windows_flat])
        new_features[f"{indicator}_SOM"] = np.pad(som_features[:, 0] * 10 + som_features[:, 1], (window_size-1, 0), 'constant')

        # GAN, RBFN, DBN placeholders
        new_features[f"{indicator}_GAN"] = np.nan
        new_features[f"{indicator}_RBFN"] = np.nan
        new_features[f"{indicator}_DBN"] = np.nan

    return new_features
```

### Final Integration

After generating the features, integrate them with the original dataset:

```python
indicators = ['MACD', 'ADX', 'RSI', 'Stochastic', 'ATR', 'OBV']  # Example indicators
new_features_df = create_indicator_features(df, indicators)
df_with_features = pd.concat([df, new_features_df], axis=1)
```

## Summary

This process generates a rich set of features for each indicator by applying various neural network transformations:

- **CNN, LSTM, RNN, MLP**: Sequence-based models to capture temporal dependencies.
- **SOM**: Clusters similar patterns in each indicator’s windowed sequences.
- **GAN, RBFN, DBN**: Placeholder models for additional exploration.

These transformed features provide additional insights and patterns that enhance the predictive power for downstream analysis.

````python
def standardize_columns(df):
    """
    Ensures that the DataFrame has columns named 'Open', 'High', 'Low', 'Close', and 'Volume'.
    If the columns are lowercase or differently named, they will be renamed accordingly.
    """
    # Define expected column mappings for both lowercase and uppercase
    expected_columns = {
        'datetime': 'Datetime',
        'open': 'Open',
        'high': 'High',
        'low': 'Low',
        'close': 'Close',
        'volume': 'Volume',
        'adj close': 'Adj Close',
        'ticker': 'Ticker',
    }
    
    # Create a mapping for renaming based on the existing columns in df
    rename_columns = {col: expected_columns[col.lower()] for col in df.columns if col.lower() in expected_columns}
    
    # Rename the columns to match the expected format
    df = df.rename(columns=rename_columns)
    
    # Verify if required columns are present after renaming
    missing_columns = [col for col in ['Open', 'High', 'Low', 'Close', 'Adj Close'] if col not in df.columns]
    if missing_columns:
        raise ValueError(f"Data is missing required columns: {missing_columns}")

    return df
````

````python
def indicator_middleware(indicator_name):
    # Check if the indicator is in pandas-ta
    if hasattr(ta, indicator_name):
        indicator_func = getattr(ta, indicator_name)
        return indicator_func

    # Check if the indicator is in ta-lib (handling candlestick patterns and standard indicators)
    elif hasattr(talib, f"CDL{indicator_name.upper()}") or hasattr(talib, indicator_name.upper()):
        if hasattr(talib, f"CDL{indicator_name.upper()}"):
            indicator_func = getattr(talib, f"CDL{indicator_name.upper()}")
        else:
            indicator_func = getattr(talib, indicator_name.upper())
        return indicator_func
    
    else:
        print(f"Indicator '{indicator_name}' is a placeholder or not found in either library.")
        return None

# Example usage
# rsi=indicator_middleware('rsi')
````

````python
def calculate_indicators(df):
    # Ensure 'Close', 'High', 'Low', 'Volume' columns are present
    required_columns = ['Close', 'High', 'Low', 'Volume']
    for col in required_columns:
        if col not in df.columns:
            raise ValueError(f"DataFrame must contain '{col}' column")

    # Trend Indicators
    df['MACD'] = ta.macd(df['Close']).iloc[:, 0]
    df['ADX'] = ta.adx(df['High'], df['Low'], df['Close'])['ADX_14']
    df['Parabolic_SAR'] = ta.psar(df['High'], df['Low'], df['Close'])['PSARl_0.02_0.2']

    # Candlestick Patterns (these are complex, so we’ll leave them as summaries)
    df['Doji'] = ta.cdl_doji(df['Open'], df['High'], df['Low'], df['Close'])
    df['Engulfing'] = ta.cdl_doji(df['Open'], df['High'], df['Low'], df['Close'])
    df['Hammer'] = ta.cdl_hammer(df['Open'], df['High'], df['Low'], df['Close'])

    # Cycle Indicators
    df['Elliott_Wave'] = ta.ewo(df['Close'])
    # Fourier Transform Analysis and Seasonal Analysis are typically more custom/complex to implement in `pandas-ta`

    # Momentum Indicators
    df['RSI'] = ta.rsi(df['Close'])
    df['Stochastic_Oscillator'] = ta.stoch(df['High'], df['Low'], df['Close'])['STOCHk_14_3_3']
    df['Williams_%R'] = ta.willr(df['High'], df['Low'], df['Close'])

    # Overlap Indicators
    df['SMA_50'] = ta.sma(df['Close'], length=50)
    df['Bollinger_High'], df['Bollinger_Mid'], df['Bollinger_Low'] = ta.bbands(df['Close'])[['BBU_20_2.0', 'BBM_20_2.0', 'BBL_20_2.0']].T.values
    df['Ichimoku_Cloud'] = ta.ichimoku(df['High'], df['Low'], df['Close'])['ISA_9']

    # Performance Metrics
    # Alpha, Sharpe Ratio, and Sortino Ratio require returns data rather than OHLCV, so these are omitted here.

    # Statistical Indicators
    # Correlation and Beta would be handled in broader context, often by comparing with a benchmark

    # Utility Indicators
    df['Fibonacci_Retracement'] = ta.fib(df['Close'])
    df['Pivot_Points'] = ta.pivot(df['High'], df['Low'], df['Close'])['P']
    df['VWAP'] = ta.vwap(df['High'], df['Low'], df['Close'], df['Volume'])

    # Volatility Indicators
    df['ATR'] = ta.atr(df['High'], df['Low'], df['Close'])
    df['Volatility_Bands_High'], df['Volatility_Bands_Low'] = df['Bollinger_High'], df['Bollinger_Low']  # reuse Bollinger bands for volatility

    # Volume Indicators
    df['OBV'] = ta.obv(df['Close'], df['Volume'])
    df['Volume_Oscillator'] = ta.pvo(df['Volume'])['PVO_12_26']
    df['Chaikin_Money_Flow'] = ta.cmf(df['High'], df['Low'], df['Close'], df['Volume'])

    return df
````
