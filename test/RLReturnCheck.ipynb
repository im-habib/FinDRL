```python
import pandas as pd
import pandas_ta as ta
```

```python
class TechnicalIndicators:
    def __init__(self, data):
        """Initialize with a DataFrame containing OHLCV data"""
        self.data = data

    def calculate_macd(self):
        """Calculate MACD and save to DataFrame"""
        macd = ta.macd(self.data['close'])
        self.data['macd'] = macd['MACD_12_26_9']
        self.data['macd_signal'] = macd['MACDs_12_26_9']
        self.data['macd_hist'] = macd['MACDh_12_26_9']

    def calculate_bollinger_bands(self):
        """Calculate Bollinger Bands and save to DataFrame"""
        bollinger = ta.bbands(self.data['close'], length=20)
        self.data['boll_ub'] = bollinger['BBU_20_2.0']
        self.data['boll_lb'] = bollinger['BBL_20_2.0']

    def calculate_rsi(self, period=30):
        """Calculate RSI with a specified period and save to DataFrame"""
        self.data['rsi_30'] = ta.rsi(self.data['close'], length=period)

    def calculate_cci(self, period=30): # DNN ---> 1 - 100000:
        """Calculate CCI with a specified period and save to DataFrame"""
        self.data['cci_30'] = ta.cci(self.data['high'], self.data['low'], self.data['close'], length=period)

# 100 indicators for buy / sell --> action
    
    # def calculate_dx(self, period=30):
    #     """Calculate DX with a specified period and save to DataFrame"""
    #     self.data['dx_30'] = ta.dx(self.data['high'], self.data['low'], self.data['close'], length=period)
    
    def calculate_dx(self, period=30):
        """Calculate DX (Directional Movement Index) with a specified period and save to DataFrame"""
        # ADX function provides both ADX and DX values
        adx_data = ta.adx(self.data['high'], self.data['low'], self.data['close'], length=period)
        self.data['dx_30'] = adx_data[f'DMP_{period}'] - adx_data[f'DMN_{period}']  # Alternative DX calculation
        self.data['adx_30']
    
    
    def calculate_sma(self, period=30, column_name='close_30_sma'):
        """Calculate Simple Moving Average (SMA) with a specified period and save to DataFrame"""
        self.data[column_name] = ta.sma(self.data['close'], length=period)

    def calculate_all_indicators(self):
        """Calculate all required indicators and save to DataFrame"""
        self.calculate_macd()
        self.calculate_bollinger_bands()
        self.calculate_rsi(period=30)
        self.calculate_cci(period=30)
        # self.calculate_dx(period=30)
        # self.calculate_sma(period=30, column_name='close_30_sma')
        # self.calculate_sma(period=60, column_name='close_60_sma')

    def get_data(self):
        """Return the DataFrame with all calculated indicators"""
        return self.data
```

```python
data = pd.read_csv("./data/usa/processed.csv")

data
```

```python

# # Initialize the TechnicalIndicators class
indicator_calculator = TechnicalIndicators(data)

# # Calculate all indicators
indicator_calculator.calculate_all_indicators()

# # Get the data with indicators
result = indicator_calculator.get_data()
print(result)
```
