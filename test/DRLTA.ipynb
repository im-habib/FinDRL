```python
import itertools

# Generate all parameter combinations for each indicator
def generate_param_combinations(bounds):
    keys, ranges = zip(*bounds.items())
    for combination in itertools.product(*[range(start, end + 1) for start, end in ranges]):
        yield dict(zip(keys, combination))
```
